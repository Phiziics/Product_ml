# Uplift Model Evaluation

## Introduction

The previous notebook built a customer-level uplift model using a Logistic Regression T-Learner.

That model estimated:

> How much each customer's conversion probability changes because of receiving the Mens E-Mail campaign.

However, generating uplift scores is not enough.

The model must demonstrate that customers ranked as high uplift actually produce more incremental conversions than customers selected randomly.

This notebook evaluates the uplift model using treatment-effect-specific metrics.

## Main Questions

1. Does the model rank high-response customers above low-response customers?
2. Does targeted treatment outperform random treatment?
3. How much incremental value is captured at different targeting depths?
4. Is the uplift ranking stable enough to support business decisions?

## Workflow

Predicted uplift scores  
→ rank customers  
→ cumulative treatment/control outcomes  
→ uplift curve  
→ Qini curve  
→ random baseline  
→ AUUC  
→ Qini coefficient  
→ targeting-depth analysis  
→ business recommendation
